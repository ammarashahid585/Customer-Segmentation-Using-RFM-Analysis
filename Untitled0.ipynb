```python
# RFM Segmentation - Colab ready
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (8,5)

# 1) Upload dataset (Excel or CSV)
print("Upload Online Retail file (Online Retail.xlsx or CSV)...")
uploaded = files.upload()
fn = list(uploaded.keys())[0]
print("Uploaded file:", fn)

# 2) Read file based on extension
if fn.lower().endswith(('.xls', '.xlsx')):
    df = pd.read_excel(fn)
else:
    # many versions use latin1 encoding
    df = pd.read_csv(fn, encoding='ISO-8859-1')

# 3) Quick inspect
print("\nColumns:", df.columns.tolist())
print("Shape:", df.shape)
display(df.head())

# 4) Data cleaning
# Drop rows without CustomerID
if 'Customer ID' not in df.columns:
    raise ValueError("Customer ID column not found — check dataset.")
df = df.dropna(subset=['Customer ID']).copy()
df['Customer ID'] = df['Customer ID'].astype(int)

# Convert InvoiceDate
if 'InvoiceDate' not in df.columns:
    raise ValueError("InvoiceDate column not found — check dataset.")
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Remove cancelled invoices (InvoiceNo starting with 'C')
if 'Invoice' in df.columns:
    df = df[~df['Invoice'].astype(str).str.startswith('C')]

# Remove negative quantity or non-positive prices if present
if 'Quantity' in df.columns and 'Price' in df.columns:
    df = df[(df['Quantity'] > 0) & (df['Price'] > 0)]
else:
    raise ValueError("Quantity or Price column missing — check dataset.")

# 5) Create TotalPrice
df['TotalPrice'] = df['Quantity'] * df['Price']

# 6) RFM aggregation
snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)
rfm = df.groupby('Customer ID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'Invoice': pd.Series.nunique,
    'TotalPrice': 'sum'
}).reset_index()
rfm.columns = ['Customer ID', 'Recency', 'Frequency', 'Monetary']
rfm['Monetary'] = rfm['Monetary'].round(2)

print("\nSample RFM (first 5):")
display(rfm.head())

# 7) RFM scoring (quartiles). Use try/except because qcut can fail with few unique values.
def safe_qcut(series, labels, ascending=True):
    try:
        if ascending:
            return pd.qcut(series, 4, labels=labels).astype(int)
        else:
            return pd.qcut(series, 4, labels=labels).astype(int)
    except Exception as e:
        # fallback to cut
        bins = 4
        return pd.cut(series.rank(method='first'), bins, labels=labels).astype(int)

# For Recency: lower is better → assign labels so smallest recency gets 4
rfm['R_score'] = safe_qcut(rfm['Recency'], labels=[4,3,2,1])

# For Frequency and Monetary: higher is better → smallest label 1, highest label 4
rfm['F_score'] = safe_qcut(rfm['Frequency'].rank(method='first'), labels=[1,2,3,4])
rfm['M_score'] = safe_qcut(rfm['Monetary'], labels=[1,2,3,4])

# In case safe_qcut returned NA, fill with 1
rfm[['R_score','F_score','M_score']] = rfm[['R_score','F_score','M_score']].fillna(1).astype(int)

# 8) Combined score and segment
rfm['RFM_Score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)
rfm['RFM_numeric'] = rfm['R_score']*100 + rfm['F_score']*10 + rfm['M_score']

def label_segment(row):
    R, F, M = row['R_score'], row['F_score'], row['M_score']
    if R >= 3 and F >= 3 and M >= 3:
        return 'Champions'
    if R >= 3 and F >= 3:
        return 'Loyal Customers'
    if F >= 3 and M >= 3:
        return 'Frequent High-Value'
    if R >= 3 and M >= 3:
        return 'Recent Big Spenders'
    if R == 4:
        return 'Recent Customers'
    if R <= 2 and F <= 2 and M <= 2:
        return 'At Risk / Lost'
    return 'Need Attention'

rfm['Segment'] = rfm.apply(label_segment, axis=1)

# 9) Show segment counts and basic stats
print("\nSegment counts:")
print(rfm['Segment'].value_counts())

seg_stats = rfm.groupby('Segment').agg({
    'Recency':'mean',
    'Frequency':'mean',
    'Monetary':'mean',
    'Customer ID':'count'
}).rename(columns={'Customer ID':'Count'}).sort_values('Count', ascending=False)
display(seg_stats.round(1))

# 10) Visualizations
# Segment counts
plt.figure(figsize=(8,5))
order = rfm['Segment'].value_counts().index
sns.countplot(data=rfm, y='Segment', order=order)
plt.title('Customer Segment Counts')
plt.xlabel('Number of Customers')
plt.ylabel('')
plt.tight_layout()
plt.show()

# Heatmap of average RFM by segment
plt.figure(figsize=(8,5))
sns.heatmap(seg_stats[['Recency','Frequency','Monetary']], annot=True, fmt='.1f')
plt.title('Average RFM by Segment')
plt.show()

# Top 10 customers by combined numeric RFM
print("\nTop 10 customers (highest RFM_numeric):")
display(rfm.sort_values('RFM_numeric', ascending=False).head(10))

# 11) Save results and download
out_fn = 'rfm_segments.csv'
rfm.to_csv(out_fn, index=False)
print(f"\nSaved RFM segments to {out_fn}. Downloading...")
files.download(out_fn)
```
